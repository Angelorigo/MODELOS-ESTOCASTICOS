```python
import random
import matplotlib.pyplot as plt
```

Aquí definimos lo que vamos a usar para generar la caminata aleatoria.

```python
print(f"Hola, ingrese un valor para P en el intervalo [0, 1], el cuál será la probabilidad de que dé un salto a la derecha en la recta de los enteros.")
P = float(input())
```

Aquí se pide al usuario ingresar la probabilidad deseada para P de sumar 1 a la recta de los enteros iniciada en 0.

```python
def simulacion_n_tiros(n):
  posicion = 0
  posiciones = [posicion]
  derechas = 0

  for i in range(1, n+1):
    x = random.random()
    if x <= P:
      posicion += 1
      derechas += 1
    else:
      posicion -= 1

    posiciones.append(posicion)

  izquierdas = n-derechas
```

Este sería el cuerpo del programa, en donde usé exactamente el mismo código que para generar *n volados de moneda*, usando el random, que se fija en el valor  P dado por el usuario, por lo que el valor de Q sería 1-P, posteriormente, entramos en un loop que simula este *volado* y va guardando la posición de nuestra "ficha" iniciada en 0.

```python
print(f"Resultados tras {n} tiros:")
print(f"Saltos a la derecha: {derechas} (Probabilidad: {izquierdas/n:.4f})")

if {posicion} == 0:
  print(f"Nos alejamos {posicion} unidades del origen")
```

En esta parte no hay mucho que explicar, aquí imprime los resultados obtenidos al final de la simulación.

```python
plt.figure(figsize=(10,5))
plt.plot(range(n+1), posiciones, linewidth = 1)
plt.axhline(0, color = 'black', linestyle = '--', linewidth = 1)
plt.xlabel("Número de tiros")
plt.ylabel("Posicion de la ficha en la recta de los enteros")
plt.title("CAMINATA ALEATORIA EN LA RECTA DE LOS ENTEROS INICIADA EN 0")
plt.grid(True)
plt.show
```

Todo esto se usa para graficar la simulación hecha por el loop, nombrando cada componente para una mejor comprensión.

```python
simulacion_n_tiros(1000)
```

Se hace una simulación de n tiros, que en este caso serían 1000
