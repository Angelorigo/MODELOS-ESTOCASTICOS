En el programa original, se imprime la matriz, pero esta vez vamos a obviarlo.

Continuemos ahora con el cálculo del número promedio de tiradas.

```python
M_juego = M.copy()
for j in range(21):
    M_juego[20, j] = 1 if j == 20 else 0
```

Extraemos la Matriz Q, junto con la matriz identidad necesarios para el cálculo.

```python
Q = M_juego[0:20, 0:20]
I = sp.eye(20)
```

Teniendo esto, hacemos el cálculo de la matriz fundamental $N = (I - Q)^-1$

```python
N = (I - Q).inv()
```

Cálculo del promedio de tiradas necesarias para ganar y la impresión de resultados.

```python
pasos_esperados_exactos = sum(N[0, :])
pasos_esperados_decimal = float(pasos_esperados_exactos)
print("\n=========================================================================================================\n")
print("  - ¿Cuál es el número promedio de tiradas necesarias para terminar el juego?\n")
print(f"   + Respuesta exacta: {pasos_esperados_exactos} tiradas.")
print(f"   + Respuesta en decimal: {pasos_esperados_decimal:.4f} tiradas.")
print("\n=========================================================================================================\n")
```

#**RATONCITO**

Como ya definimos la librería, vamos directo con las entradas, de igual manera, con racionales.

```python
a = sp.Rational(1, 2)
b = sp.Rational(1, 3)
c = sp.Rational(1, 4)
```

Vayamos ahora con la matriz.

```python
M = sp.Matrix([
    [0, a, a, 0, 0, 0, 0, 0, 0], # 0
    [b, 0, 0, b, 0, 0, 0, b, 0], # 1
    [b, 0, 0, b, 0, 0, 0, 0, b], # 2
    [0, c, c, 0, c, c, 0, 0, 0], # 3
    [0, 0, 0, b, 0, 0, b, b, 0], # 4
    [0, 0, 0, b, 0, 0, b, 0, b], # 5
    [0, 0, 0, 0, a, a, 0, 0, 0], # 6
    [0, 0, 0, 0, 0, 0, 0, 1, 0], # 7 Comida
    [0, 0, 0, 0, 0, 0, 0, 0, 1]  # 8 Schock
])
```

Ahora, extraigamos las matrices que se usarán para los cálculos.

```python
Q = M[0:7, 0:7]
R = M[0:7, 7:9]
I = sp.eye(7)
```

Teniendo esto, calculemos la Matriz Fundamental N = (I - Q)^-1

```python
N = (I - Q).inv()
```

Calculemos la Matriz de probabilidades de absorción B = N * R

```python
B = N * R
```

Cálculo de las probabilidades de llegar tanto a la comida, como al shock, iniciando en 0, e impresión de resultados.

```python
prob_comida = B[0, 0]
prob_toque = B[0, 1]
print("========================================================\n")
print("   RESULTADOS DEL RATONCITO:\n")
print(f"Probabilidad de llegar a la comida (casilla 7): {prob_comida}")
print(f"Probabilidad de llegar al toque  (casilla 8): {prob_toque}")
print("\n========================================================\n")
```
